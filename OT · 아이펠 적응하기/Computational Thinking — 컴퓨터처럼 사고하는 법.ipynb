## 큰 일을 통째로 던지면 왜 아무것도 안 돌아올까

---

오늘은 온보딩 모듈의 마지막 날입니다. 오늘은 AI에게 일을 시킬 때마다 꺼내 쓸 사고법을 익혀봅시다.

먼저 한 장면을 떠올려 봅시다. 어떤 사람이 ChatGPT에게 이렇게 부탁합니다. "우리 회사 업무를 AI로 자동화해 줘." 돌아오는 답은 그럴듯하지만 자기 일에는 하나도 안 맞는, 일반론으로 가득한 글입니다. 같은 사람이 다음 날엔 이렇게 부탁합니다. "고객 문의 메일을 세 종류(환불·배송·기타)로 나누고, 환불 건은 처리 기한을 함께 적어서 표로 만들어 줘." 이번엔 바로 쓸 수 있는 결과가 나옵니다.

두 부탁의 차이는 AI의 성능이 아닙니다. **일을 던지는 사람이 일을 얼마나 잘게 나눠서 줬느냐**의 차이입니다. 오늘 배우는 것이 바로 이 "나누는 사고법"입니다.

### 🎯 오늘의 핵심 한 문장

오늘 노드에서 손에 쥐고 갈 문장은 하나입니다. 나머지는 전부 이 문장을 받치는 설명입니다.

> **컴퓨테이셔널 싱킹(Computational Thinking, 컴퓨팅 사고법)이란, 큰 문제를 컴퓨터(그리고 AI)가 풀 수 있는 작은 조각으로 나누는 사고법입니다. 그리고 그 사고법이 AI에게 일을 정확히 시키는 기초입니다.**

코딩 언어가 아닙니다. **사고의 문법**입니다. 바이브 코딩 시대에는 코드를 직접 못 짜도 됩니다. 코드는 AI가 짭니다. 사람이 해야 할 일은 "무엇을, 어떤 순서로, 어떤 조건에서"를 또렷하게 나눠 주는 일입니다. 그 나누는 일이 컴퓨팅 사고법입니다.

## 하나의 사고법, 네 개의 손동작

---

컴퓨팅 사고법을 검색하면 거의 항상 네 단어가 함께 나옵니다. **분해 · 패턴 인식 · 추상화 · 알고리즘**. 그래서 "외울 게 네 개구나" 하고 부담부터 느끼기 마련입니다. 그렇게 보지 마세요.

이 넷은 별개의 개념 네 개가 아닙니다. **하나의 사고법(큰 문제를 작게 나누기)을 실제로 할 때 손이 거치는 네 동작**입니다. 요리 한 번 하는 데 "재료 손질·간 맞추기·불 조절·담기"가 따로 노는 기술이 아니라 한 번의 요리를 이루는 동작인 것과 같습니다. 네 동작을 하나씩 봅니다.

참고로 이 용어는 2006년 카네기멜런대의 자넷 윙(Jeannette Wing) 교수가 짧은 글에서 대중화했습니다(아래 '출처'). 윙은 컴퓨팅 사고법이 컴퓨터 전문가만의 것이 아니라 모든 분야의 사람에게 쓸모 있는 기본 소양이라고 봤습니다.

### 🔬 동작 1 — 분해: 통째로 막막한 것을 시작할 수 있는 조각으로

복잡한 문제 하나는 막막합니다. 작은 문제 여러 개로 쪼개면 각각은 "오늘 안에 시작할 수 있는 일"이 됩니다.

결혼식을 떠올려 봅시다. "결혼식을 준비하라"는 막막합니다. 그런데 예산 짜기 / 하객 명단 / 장소 예약 / 청첩장 / 식순 / 음식 주문 / 신혼여행으로 쪼개면, 각 항목은 지금 당장 손댈 수 있는 크기가 됩니다. 분해의 기준은 하나입니다. **"한 사람이 한 시간 안에 시작할 수 있는 크기"** 까지 쪼갰는가. 그보다 크면 다시 나눕니다.

AI에게 일을 시킬 때도 같습니다. "업무를 자동화해 줘"는 결혼식을 통째로 던진 것과 같습니다. "메일을 세 종류로 나눠 표로 만들어 줘"는 한 조각으로 쪼갠 것입니다.

### 🔬 동작 2 — 패턴 인식: 반복되는 것을 알아보기

쪼갠 조각들 사이에서 **반복·공통점**을 찾는 동작입니다. 같은 형식이 반복되면, 한 가지 해결책으로 여러 조각을 한꺼번에 풀 수 있습니다.

가계부를 정리할 때 "스타벅스 5,500원", "투썸 4,800원", "이디야 4,000원"이 반복되면 머릿속에서 자연스럽게 "카페"로 묶입니다. 한 달에 카페에 23만 원 썼다는 한 줄이 그렇게 나옵니다. 이 "묶어 보는" 동작이 패턴 인식입니다.

도메인 전문가는 이 동작을 이미 매일 합니다. 환자 100명의 증상에서, 100건의 판례에서, 100건의 캠페인 결과에서 반복을 알아보는 일 — 그게 패턴 인식입니다.

### 🔬 동작 3 — 추상화: 본질만 남기고 나머지는 버리기

패턴을 찾았으면, 그 패턴의 **본질만 남기고 나머지 세부는 과감히 버리는** 동작입니다.

지하철 노선도를 떠올려 봅시다. 실제 서울 지하철은 곡선이고 역 사이 거리도 제각각이지만, 노선도는 직선에 등간격으로 그려집니다. "어느 역에서 갈아탈 수 있는가"라는 본질만 남기고 실제 거리·곡선은 버린 것입니다. 그래서 오히려 길 찾기가 또렷해집니다.

추상화가 중요한 이유는 다음 섹션에서 보겠지만, **컴퓨터는 애매함을 못 알아듣기 때문**입니다. 사람의 모호한 표현("적당히", "심한")을 컴퓨터가 알아들을 수 있는 또렷한 형태로 옮기는 동작이 추상화입니다.

### 🔬 동작 4 — 알고리즘: 푸는 순서를 정하기

마지막은 문제를 푸는 **단계별 순서**를 정하는 동작입니다. 같은 일이라도 순서가 바뀌면 결과가 달라집니다.

빨래를 떠올려 봅시다. "흰 빨래·색 빨래 분리 → 흰 빨래 먼저 세탁 → 건조 → 개기" 순서가 있습니다. 흰 빨래와 색 빨래를 분리하는 단계를 건너뛰면 흰옷에 물이 듭니다. 순서 자체가 결과의 일부입니다.

여러분의 도메인에도 이미 알고리즘이 있습니다. 의사의 ABCDE 평가(기도·호흡·순환·의식·노출 순서), 변호사의 "사실관계 → 쟁점 → 법률 → 판례 → 결론", 마케터의 "시장 조사 → 목표 → 제작 → 실행 → 분석" — 모두 알고리즘입니다.

### 💡 네 동작은 한 사이클로 돈다

네 동작은 "분해 → 패턴 → 추상화 → 알고리즘" 한 방향으로만 흐르지 않습니다. 서로 오갑니다.

```
:        분해 ─────────► 패턴 인식
:         ▲                │
:         │                ▼
:       알고리즘 ◄───────  추상화
```

알고리즘을 짜다 보면 더 쪼개야 할 부분이 보여 다시 분해로 돌아가고, 패턴이 또렷해지면 추상화가 가벼워집니다. 네 동작 사이를 자연스럽게 오가는 것이 익숙해진 상태입니다. 한 번에 완벽하게 하려 하지 마세요. 첫 시도는 거칠고, 두세 번째에서 또렷해집니다.

## 의사도 변호사도 이미 매일 하고 있다

---

컴퓨팅 사고법이 낯선 새 기술처럼 느껴진다면, 한 가지 사실을 먼저 짚고 가야 합니다. **여러분은 이 사고법을 자기 분야에서 이미 매일 쓰고 있습니다.** 단지 "분해", "추상화" 같은 이름을 붙여 본 적이 없을 뿐입니다. 오늘의 학습은 새 능력을 배우는 게 아니라, 이미 가진 능력에 이름을 붙이고 그것을 AI에게 옮기는 일입니다.

한 도메인을 깊이 들여다봅니다. 그다음 섹션 4에서 여러분 자신의 도메인으로 옮깁니다.

### 🔬 의사가 응급실에서 첫 1분에 하는 일

응급실에 환자가 들어옵니다. 경력 있는 의사는 첫 1분 안에 머릿속에서 다음을 거칩니다.

- **분해**: 환자 상태를 기도·호흡·순환·의식·노출로 나눠 본다 (ABCDE)
- **패턴 인식**: "이건 흉통 + 식은땀 + 방사통 — 익숙한 그 패턴"
- **추상화**: 수많은 정보 중 응급도를 결정하는 핵심만 남긴다
- **알고리즘**: 어떤 검사를 어떤 순서로 할지 정한다

의사는 이 네 동작을 "컴퓨팅 사고법"이라 부르지 않습니다. 그냥 "진료"라고 부릅니다. 하지만 구조는 똑같습니다. 그리고 흥미로운 점이 있습니다. 같은 ABCDE라도 5년 차와 30년 차의 적용이 다릅니다. 30년 차는 표준 단계에 미묘한 직관과 예외 케이스를 결합합니다. **이 깊이가 AI에게 옮길 수 없는, 도메인 전문가만의 자산입니다.**

### 🔬 변호사·마케터·디자이너의 같은 네 동작

| 도메인 | 분해 | 패턴 인식 | 추상화 | 알고리즘 |
|---|---|---|---|---|
| **변호사** | 사건을 사실·쟁점·법률로 나눔 | 유사 판례를 알아봄 | 핵심 쟁점 한 줄로 | 사실→쟁점→법률→판례→결론 |
| **마케터** | 캠페인을 채널·지표·시간대로 나눔 | "주말 저녁 전환율↑" 패턴 | "이벤트=클릭률↑" 본질 | 예산 배분→실행→주간 점검 |
| **디자이너** | 화면을 헤더·CTA·푸터로 나눔 | 유사 앱의 UI 패턴 | 핵심 액션 1개만 남김 | 인터뷰→와이어프레임→테스트 |

세 도메인 모두 "코딩"과는 거리가 멀어 보이지만, 일하는 머릿속 구조는 컴퓨팅 사고법과 같습니다. 변호사의 판례 패턴 인식, 마케터의 캠페인 분류, 디자이너의 본질 추출 — 전부 이미 하고 있는 네 동작입니다.

### 🔬 그렇다면 AI는 어디에 끼어드나

2026년 현재, AI 도구는 네 동작을 "어느 정도" 대신해 줄 수 있습니다. 하지만 결정적인 한계가 있습니다.

| 동작 | 사람(도메인 전문가)의 강점 | AI의 강점 |
|---|---|---|
| 분해 | 도메인 깊이로 *진짜* 단위를 찾음 | 일반적 분해를 빠르게 생성 |
| 패턴 인식 | 미묘한 도메인 패턴을 알아봄 | 대량 데이터에서 패턴을 빠르게 추출 |
| 추상화 | 무엇이 본질인지 *판단* | 일반적 추상화를 빠르게 |
| 알고리즘 | 도메인 맥락에 맞는 순서 | 표준 순서를 빠르게 |

핵심은 이렇습니다. **AI는 네 동작을 모두 빠르게 하지만, 여러분 도메인의 진짜 깊이가 없습니다.** 도메인 맥락을 정확히 알려 주면 AI는 그 안에서 빠르게 일합니다. 안 알려 주면 일반적이고 흔한 답을 줍니다. 그래서 AI 시대일수록 도메인 깊이가 더 중요해집니다.

> 📊 **연구가 말하는 것**: 사람 단독·AI 단독·둘의 조합을 비교한 대규모 메타분석(106개 실험 종합)에 따르면, 사람과 AI의 조합이 *항상* 더 낫지는 않았습니다. 평균적으로는 '사람 혼자'나 'AI 혼자' 중 더 나은 쪽에 못 미치는 경우도 있었고, 특히 **결정·판단** 과제에서 그랬습니다. 반대로 **사람이 AI보다 나은 영역**일수록 조합의 이득이 커졌습니다(한 예로 새 사진 분류에서 사람 81%·AI 73%·조합 90%). 정리하면, 조합의 이득은 '사람이 무엇을 더 잘하느냐'에서 나옵니다. 도메인 전문가의 판단이 바로 그 '더 잘하는 부분'입니다. (Vaccaro 등 2024 — 아래 '출처')

### 💡 정직하게 — 컴퓨팅 사고법은 만능이 아닙니다

여기서 솔직하게 짚고 갑니다. 컴퓨팅 사고법으로 **모든** 문제를 풀 수 있다고 말하면 거짓말입니다. 네 동작으로 쪼개기 어려운 영역이 분명히 있습니다.

- **본질적으로 창의적인 작업** — 시·소설 쓰기, 새로운 콘셉트 발상. 분해·순서로 다 풀리지 않습니다.
- **사람 사이의 미묘한 영역** — 환자의 "괜찮아요" 한 마디를 표정·맥락에서 읽어 내는 일.
- **정답 없는 가치 판단** — 치료 우선순위, 변론 전략, 사용자에게 미칠 영향에 대한 윤리적 결정.

이 영역들은 컴퓨팅 사고법이 "한 가지 시각"일 뿐, 모든 답이 아닙니다. 그리고 바로 이 영역들이 도메인 전문가의 **사라지지 않는 자리**입니다. AI가 잘게 쪼갤 수 있는 일은 AI에게 맡기고, 사람은 이 깊은 판단으로 옮겨 갑니다.

> - AI에게 넘길 수 있는 건 *판단을 돕는 자료 정리*(예: 객관 지표를 표로 묶기)까지입니다.
> - 최종 판단·그 판단의 책임은 사람에게 남아야 합니다. "무엇을 AI에게 맡기고 무엇을 안 맡길지" 자체가 사람의 결정입니다.
>
> </details>

## 내 문제를 네 조각으로, 그리고 AI에게

---

이제 손을 움직일 시간입니다. 코드는 짜지 않습니다. 대신 **여러분 도메인의 진짜 문제 하나**를 골라 네 동작으로 분해한 한 페이지를 만듭니다. 이 한 페이지가 오늘의 산출물이고, 넉 달 학습의 출발 자료가 됩니다.

강사가 답을 주지 않습니다. 여러분 도메인을 아는 사람은 여러분뿐이기 때문입니다. 예시는 보여 드리지만, 그대로 베끼면 자기 것이 되지 않습니다.

### 📝 1단계 — 문제 한 가지 고르기

먼저 자기 도메인에서 **자주 반복되고, 시간을 많이 잡아먹는** 작업 하나를 고릅니다. "하고 싶은 멋진 일"이 아니라 "매번 하는 귀찮은 일"이 좋은 첫 후보입니다. 반복되는 일일수록 AI에게 맡기기 좋기 때문입니다.

예시:
- 의사: "외래 후 진료 기록 요약 작성"
- 변호사: "상담 들어온 사건을 유형별로 1차 분류"
- 마케터: "매주 광고 캠페인 결과를 한 장으로 정리"

여러분 문제 한 줄: `____________________`

### 📝 2단계 — 네 동작 워크북 (빈칸을 자기 도메인으로)

고른 문제를 네 동작으로 쪼개 봅니다. 아래 표의 가운데 칸은 마케터 예시입니다. 오른쪽 칸을 **자기 도메인**으로 채우세요. 처음엔 거칠어도 괜찮습니다.

| 동작 | 마케터 예시(광고 결과 정리) | 내 도메인 |
|---|---|---|
| **분해** (5~7개 작은 단위로) | 채널별·캠페인별·지표별·시간대별로 나눔 | `______` |
| **패턴 인식** (반복 2~3개) | 주말 저녁 전환율↑ / 이벤트 클릭률↑ | `______` |
| **추상화** (본질 한 줄) | "한 줄 인사이트 3개로 한 달을 요약" | `______` |
| **알고리즘** (단계별 순서) | 입력→4가지 분리→패턴 추출→인사이트 출력 | `______` |

> 빈칸을 *자기* 업무로 채우세요. 가장 어려운 칸이 보일 겁니다. 그 칸이 섹션 2에서 적어 둔 "가장 어색한 동작"과 같다면, 그게 넉 달 동안 가장 많이 자라날 부분입니다.

### 📝 3단계 — 네 동작 분석을 AI 지시로 옮기기

네 동작으로 쪼갠 결과가 있으면, 그것을 그대로 AI에게 전달할 수 있는 형식으로 옮깁니다. 이것이 다음 모듈(바이브 코딩)에서 매일 하게 될 일의 예고편입니다.

아래는 마케터의 전달 형식입니다. 한 단계라도 빠지면 AI가 빈 칸을 자기 추측으로 채워 결과가 흔들립니다.

```
[도메인 맥락]
나는 광고 대행사 마케터 7년 차이고, 매주 캠페인 결과를 한 장으로 정리하고 싶습니다.

[네 동작 분석]
1. 분해: 채널별·캠페인별·지표별·시간대별 데이터로 나눔
2. 패턴: 시간대 효과, 캠페인 종류 효과, 채널 효과 3가지
3. 추상화: 한 줄 인사이트 3개로 한 달 요약
4. 알고리즘: 데이터 입력 → 4가지 분리 → 패턴 추출 → 인사이트 3개 출력

[출력 요구]
클라이언트 보고서 양식으로, 한국어로.

[검증 기준]
시니어 마케터 2명이 "실제로 쓸 수 있다"고 평가할 것.
```

이 형식에서 가장 중요한 칸은 맨 위 **[도메인 맥락]**입니다. 자기가 누구이고 무엇을 풀려는지를 한 단락으로 적는 일이 AI 응답의 질을 가장 크게 좌우합니다.

### 🎮 직접 해보기 — 자기 도메인 전달 형식 채우기

위 마케터 형식을 자기 도메인으로 바꿔 메모장이나 노션에 적어 보세요. 빈칸 형식만 옮겨 둡니다.

```
[도메인 맥락]
나는 ____ 도메인 ____ 년 차이고, ____ 작업을 자동화/정리하고 싶습니다.

[네 동작 분석]
1. 분해: ____
2. 패턴: ____
3. 추상화: ____
4. 알고리즘: ____

[출력 요구]
____ 형식으로, 한국어로.

[검증 기준]
____ 한 사람이 ____ 라고 평가할 것.
```

> 🔍 **자기 점검**: 위 형식에서 한 칸이라도 비어 있다면, 그 칸은 "아직 내가 또렷하게 생각하지 못한 부분"입니다. 빈 칸 자체가 내일 무엇을 더 다듬어야 할지 알려 주는 신호입니다.
>
> <details><summary>💡 자주 빠지는 칸 — 예시 답안</summary>
>
> 가장 자주 비는 칸은 **[검증 기준]**입니다. "잘 작동하면 좋겠어"로 끝내면 AI가 자기 기준으로 만들어 검증이 어렵습니다. "누가", "어떤 기준으로" 합격을 판단하는지 미리 적으면 AI가 그 기준에 맞춰 결과를 만듭니다. 검증 기준은 도메인 전문가만 정할 수 있는 칸입니다.
>
> </details>
